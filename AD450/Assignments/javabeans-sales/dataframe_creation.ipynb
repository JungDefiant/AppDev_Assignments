```python
import pandas as pd 

dataframe = pd.DataFrame({
    "Americano": [5628, 6239],
    "Latte": [12426, 25384],
    "Espresso": [852, 756],
    "Cappuccino": [8522, 7561],
}, index=["2021", "2022"] )
dataframe.index.name = "Sales by Year"

dataframe
```

```python
dataframe.loc["Total"] = dataframe.sum()

dataframe
```

```python
dataframe.loc["YoY Growth"] = (dataframe.loc["2022"] - dataframe.loc["2021"]) / dataframe.loc["2021"] * 100

dataframe
```

The lowest selling beverage type between 2021 and 2022 was Cappuccino. The highest selling beverage type during this same period was Latte.

*Which products deserve higher marketing spend? Are there any drinks that might need to be phased out or redesigned based on their YoY trajectory?*

The products that deserve higher marketing spend are lattes because of their massive year-over-year growth. It makes the most sense to capitalize on this growth and "ride the wave". Americanos may not be worth the extra investment since they're growing so slowly. The drinks that likely need to be phased out or redesigned are espressos and cappuccinos, because of their declining YoY trajectory.

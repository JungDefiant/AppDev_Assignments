#### Data Quality Assessment
*Issue 1:* In imports, there is a row for Belgium, Luxembourg, and then Belgium/Luxembourg. This combined statistic of both countries can change the max and mean of each row significantly, which means a less accurate accounting of the data.

*Issue 2:* In exports, Brazil has missing values for 2014, 2015, and 2019. These missing values can cause calculation errors when processing these statistics if not accounted for.

*Issue 3:* Several entries for certain countries have 0's for multiple years. Presumably this means the country was not exporting in those years. These can skew the minimum and mean values, which can affect calculations that depend on those values.
